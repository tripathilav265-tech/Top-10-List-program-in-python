```python
prices = [120, 450, 80, 230]# max ,min,sum
print("Total sum:", sum(prices))
print("Minimum price:", min(prices))
print("Maximum price:", max(prices))
```

```python
vals = [1, 2, 3, 2, 4, 2, 5] # count and index
print("Count of 2:", vals.count(2))
print("First index of 4:", vals.index(4))
```

```python
squares = [x**2 for x in range(1, 6)] # List Comprehension
print("Squares list:", squares)
```

```python
chars = ['a', 'b', 'c', 'd']#Reversing a List
chars.reverse()
print("Reversed list:", chars)
```

```python
scores = [45, 12, 89, 5, 23]#ascending) and descending
scores.sort()
print("Ascending:", scores)
scores.sort(reverse=True)
print("Descending:", scores)
```

```python
items = [100, 200, 300, 400]#Finding Length of a List
print("Total elements in list:", len(items))
```

```python
nums = [10, 20, 30, 40, 50, 60]#list slicing
print("Slice [1:4]:", nums[1:4])
print("First three items:", nums[:3])
print("Every second item:", nums[::2])
```

```python
languages = ['Python', 'Java', 'C++', 'Ruby']#Removing Elements from a List
languages.remove('Java')
popped_item = languages.pop()
print("List after removal:", languages)
print("Removed item:", popped_item)
```

```python
numbers = [1, 2, 3]#(Adding Elements to a List
numbers.append(5)
numbers.insert(3, 4)
print("Updated list:", numbers)
```

```python
fruits = ['apple', 'banana', 'cherry', 'orange']#Creating a List and Accessing Elements
print("First element:", fruits[0])
print("Last element:", fruits[-1])
```
